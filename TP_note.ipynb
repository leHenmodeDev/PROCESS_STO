```python
import numpy as np
import matplotlib.pyplot as plt

def acf(x, l):
    n = len(x)
    m = np.mean(x)
    v = np.var(x)
    return np.mean((x[:n-l]-m)*(x[l:]-m))/v

def simuler_poisson(lmbda, T):
    t = 0
    sauts = []
    while t < T:
        t += np.random.exponential(1/lmbda)
        if t < T:
            sauts.append(t)
    return np.array(sauts)
```
